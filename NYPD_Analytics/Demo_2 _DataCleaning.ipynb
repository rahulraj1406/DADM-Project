```python
# Count duplicates
df.duplicated().sum()

# Remove duplicates
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

No duplicates

```python
df["BORO_NM"].unique()
```

```python
# Ensure numeric
df["Latitude"] = pd.to_numeric(df["Latitude"], errors="coerce")
df["Longitude"] = pd.to_numeric(df["Longitude"], errors="coerce")

# Impossible coordinates mask
invalid_coords = (
    (df["Latitude"] < 40) | (df["Latitude"] > 42) |
    (df["Longitude"] < -75) | (df["Longitude"] > -72)
)

# Count
num_invalid = invalid_coords.sum()

print("Number of impossible coordinates:", num_invalid)
```

```python
# Remove invalid lat/long if present
if "Latitude" in df.columns:
    df = df[(df["Latitude"].between(40, 42))]

if "Longitude" in df.columns:
    df = df[(df["Longitude"].between(-75, -72))]
```

```python
df.shape
```

```python
# Create Consistent Time Features
df["year"] = df["CMPLNT_FR_DT"].dt.year.astype("Int64")
df["month"] = df["CMPLNT_FR_DT"].dt.month.astype("Int64")
df["day"] = df["CMPLNT_FR_DT"].dt.day.astype("Int64")
df["day_of_week"] = df["CMPLNT_FR_DT"].dt.day_name()
```

```python
df.head()
```

```python
# Inconsistent Complaint Dates
if "CMPLNT_TO_DT" in df.columns:
    duration = (df["CMPLNT_TO_DT"] - df["CMPLNT_FR_DT"]).dt.days

    inconsistent_mask = duration < 0

    print("Records with inconsistent complaint dates:", inconsistent_mask.sum())

    # Drop inconsistent records
    df = df[~inconsistent_mask]

    print("Remaining rows after removal:", df.shape[0])
```

```python
# Remove irrelevant columns
df = df.drop(columns=["TRANSIT_DISTRICT", "HOUSING_PSA", "PD_CD", "CMPLNT_TO_DT", "CMPLNT_TO_TM"])
```

```python
# Removing rows with missing values in key attributes (borough, offense description, complaint date)before = df.shape[0]

df = df.dropna(subset=["BORO_NM", "OFNS_DESC", "CMPLNT_FR_DT"])

after = df.shape[0]

print("Rows dropped:", before - after)
print("Remaining rows:", after)
```
